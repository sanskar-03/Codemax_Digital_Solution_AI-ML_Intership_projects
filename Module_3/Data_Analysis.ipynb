```python
# Import the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load a sample dataset (We'll use a public dataset of Titanic passengers)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Data loaded successfully!")
```

```python
# 1. Explore the dataset
print("--- First 5 rows of data ---")
display(df.head())

print("\n--- Dataset Information ---")
df.info()

# 2. Simple Data Cleaning
# Let's fill missing 'Age' values with the average (mean) age
mean_age = np.mean(df['Age'])
df['Age'] = df['Age'].fillna(mean_age)

# Drop any remaining rows that still have missing data
df_cleaned = df.dropna()

print("\nData cleaning complete. No more missing values!")
```

```python
# 1. Bar Chart: Count of passengers by class
plt.figure(figsize=(8, 4))
df_cleaned['Pclass'].value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title('Bar Chart: Passengers per Class')
plt.xlabel('Ticket Class (1, 2, 3)')
plt.ylabel('Number of Passengers')
plt.show()

# 2. Line Chart: Average fare paid by passenger class
plt.figure(figsize=(8, 4))
df_cleaned.groupby('Pclass')['Fare'].mean().plot(kind='line', marker='o', color='green')
plt.title('Line Chart: Average Fare by Class')
plt.xlabel('Ticket Class')
plt.ylabel('Average Fare')
plt.grid(True)
plt.show()

# 3. Pie Chart: Survival rate (0 = No, 1 = Yes)
plt.figure(figsize=(6, 6))
df_cleaned['Survived'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['red', 'lightgreen'])
plt.title('Pie Chart: Survival Rate')
plt.ylabel('') # Hides the y-axis label for a cleaner pie chart
plt.show()
```
